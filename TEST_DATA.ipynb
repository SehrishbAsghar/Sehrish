```python
import os, subprocess
if not os.path.exists("HIGGS.csv"):
    subprocess.run("wget -q https://archive.ics.uci.edu/static/public/280/higgs.zip -O higgs.zip "
                   "&& unzip -q higgs.zip && gunzip -f HIGGS.csv.gz", shell=True, check=True)
```

```python
from pyspark.sql import SparkSession

spark = (SparkSession.builder
         .appName("higgs_check")
         .master("local[*]")
         .config("spark.driver.memory", "4g")
         .getOrCreate())

df = spark.read.csv("HIGGS.csv", inferSchema=True)
df = df.toDF(*(["label"] + [f"f{i}" for i in range(1, 29)]))

print("Rows:", df.count())          # expect 11000000
print("Columns:", len(df.columns))  # expect 29
df.select("label", "f1", "f2").show(3)
```
